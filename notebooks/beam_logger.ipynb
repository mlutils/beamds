```python
from beam import logger
from beam import resource
```

Log to the screen and the default output

```python
logger.info('hi there')
```

Add handler to a local directory, where logs will be appended to a single file

```python
logger.add_file_handlers('/tmp/log_dir')
```

```python
logger.info('hi there')
```

```python
!cat /tmp/log_dir/debug.log
```

Add handler to a local directory, where each log will be saved in a separate file

```python
logger.add_directory_handler('/tmp/log_dir2', serialize=True)
```

```python
logger.warning('Each log will be saved in a separate file, can log also extra data', some_exra_data=123)
```

```python
!ls -la /tmp/log_dir2
```

```python
!cat /tmp/log_dir2/20250421-123642-055_969391_WARNING_1262031823.py__module__1.json
```

Can also use elasticsearch to log the files

```python
es = resource('elastic://84.108.252.235:31063')
```

```python
es = resource('elastic://84.108.252.235:31063')
```

```python
logger.add_directory_handler(es.joinpath('logs'), serialize=True)
```
